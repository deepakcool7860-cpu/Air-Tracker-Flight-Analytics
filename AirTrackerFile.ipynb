```python
def extract_flight_info(flight_data, flight_type, airport_icao):
    flight_id = flight_data.get('number', '') + '_' + flight_data.get(flight_type, {}).get('scheduledTime', {}).get('utc', '')

    # Determine origin and destination based on flight type, ensuring ICAO codes
    origin_code = ''
    destination_code = ''
    if flight_type == 'departure':
        origin_code = airport_icao # The airport from which the flights were fetched
        destination_code = flight_data.get('arrival', {}).get('airport', {}).get('icao')
    elif flight_type == 'arrival':
        origin_code = flight_data.get('departure', {}).get('airport', {}).get('icao')
        destination_code = airport_icao # The airport to which the flights were fetched

    return {
        'flight_id': flight_id,
        'flight_number': flight_data.get('number'),
        'aircraft_registration': flight_data.get('aircraft', {}).get('reg'),
        'origin_icao': origin_code,
        'destination_icao': destination_code,
        'scheduled_departure': flight_data.get('departure', {}).get('scheduledTime', {}).get('local'),
        'actual_departure': flight_data.get('departure', {}).get('revisedTime', {}).get('local', flight_data.get('departure', {}).get('scheduledTime', {}).get('local')),
        'scheduled_arrival': flight_data.get('arrival', {}).get('scheduledTime', {}).get('local'),
        'actual_arrival': flight_data.get('arrival', {}).get('revisedTime', {}).get('local', flight_data.get('arrival', {}).get('scheduledTime', {}).get('local')),
        'status': flight_data.get('status'),
        'airline_code': flight_data.get('airline', {}).get('iata')
    }

all_flights_data = []

flight_api_response = flight_response.json()

# Extract departures
if 'departures' in flight_api_response:
    for flight in flight_api_response['departures']:
        all_flights_data.append(extract_flight_info(flight, 'departure', 'VIDP')) # 'VIDP' is the ICAO for the Delhi airport used in the API call

# Extract arrivals
if 'arrivals' in flight_api_response:
    for flight in flight_api_response['arrivals']:
        all_flights_data.append(extract_flight_info(flight, 'arrival', 'VIDP')) # 'VIDP' is the ICAO for the Delhi airport used in the API call

# Create DataFrame
flights_df = pd.DataFrame(all_flights_data)
display(flights_df)
```

```python
cursor = connection.cursor()

# 1) Create a database (skip if it already exists)
cursor.execute("CREATE DATABASE IF NOT EXISTS AirTracker_db")
print("Database created or already exists.")

# 2) Switch to that database
cursor.execute("USE AirTracker_db")

# 3) Create a table
create_table_query = '''
CREATE TABLE IF NOT EXISTS airport (
    airport_id INT AUTO_INCREMENT PRIMARY KEY,
    icao_code  VARCHAR(4)   UNIQUE,
    iata_code  VARCHAR(3)   UNIQUE,
    name       VARCHAR(255),
    city       VARCHAR(100),
    country    VARCHAR(100),
    continent  VARCHAR(50),
    latitude   DECIMAL(9,6),
    longitude  DECIMAL(9,6),
    timezone   VARCHAR(64)
)
'''
cursor.execute(create_table_query)
connection.commit()   # commit() saves the change permanently — like clicking the "Apply" button in Workbench
print("Table 'airport' created or already exists.")
```

```python
cursor.execute("USE AirTracker_db")

create_table_query = '''
CREATE TABLE IF NOT EXISTS aircraft (
    aircraft_id     INT NOT NULL AUTO_INCREMENT,
    registration    VARCHAR(20) UNIQUE,
    model           VARCHAR(100),
    manufacturer    VARCHAR(100),
    icao_type_code  VARCHAR(4),
    owner           VARCHAR(150),
    PRIMARY KEY (aircraft_id)
)
'''

cursor.execute(create_table_query)
connection.commit()

print("Table 'aircraft' created or already exists.")
```

```python
cursor.execute("USE AirTracker_db")

create_table_query = '''
CREATE TABLE IF NOT EXISTS flights (
    flight_id             VARCHAR(100) NOT NULL,
    flight_number         VARCHAR(20),
    aircraft_registration VARCHAR(20),
    origin_iata           VARCHAR(3),
    destination_iata      VARCHAR(3),
    scheduled_departure   DATETIME,
    actual_departure      DATETIME,
    scheduled_arrival     DATETIME,
    actual_arrival        DATETIME,
    status                VARCHAR(30),
    airline_code          VARCHAR(3),
    airline_name          VARCHAR(100),
    departure_terminal    VARCHAR(10),
    departure_gate        VARCHAR(10),
    arrival_terminal      VARCHAR(10),
    arrival_gate          VARCHAR(10),
    codeshare_status      VARCHAR(30),
    is_cargo              TINYINT(1) NOT NULL DEFAULT 0,
    distance_km           DOUBLE,
    PRIMARY KEY (flight_id),
    FOREIGN KEY (aircraft_registration) REFERENCES aircraft(registration),
    FOREIGN KEY (origin_iata)           REFERENCES airport(iata_code),
    FOREIGN KEY (destination_iata)      REFERENCES airport(iata_code)
)
'''

cursor.execute(create_table_query)
connection.commit()

print("Table 'flights' created or already exists.")
```

```python
cursor.execute("USE AirTracker_db")

create_table_query = '''
CREATE TABLE IF NOT EXISTS airport_delays (
    delay_id          INT NOT NULL AUTO_INCREMENT,
    airport_iata      VARCHAR(3),
    delay_date        DATE,
    total_flights     INT,
    delayed_flights   INT,
    delay_index       DOUBLE,
    median_delay_min  INT,
    canceled_flights  INT,
    PRIMARY KEY (delay_id),
    UNIQUE KEY uq_airport_day (airport_iata, delay_date),
    FOREIGN KEY (airport_iata) REFERENCES airport(iata_code)
)
'''

cursor.execute(create_table_query)
connection.commit()

print("Table 'airport_delays' created or already exists.")
```
